### 게임1. 레고 자르기 (13슬라이드)
- AI는 문장을 받으면 가장 먼저 무엇을 할까?
- AI의 시선으로 문장 바라보기

```
pip install tiktoken
```

```
import tiktoken   # 문장을 AI가 읽는 조각(토큰)으로 잘라주는 도구

# GPT-4가 문장을 자를 때 쓰는 '자르는 규칙'을 불러와요
enc = tiktoken.get_encoding("cl100k_base")


# 문장을 넣으면 AI가 그 문장을 어떻게 조각내서 읽는지 보여주는 기능
def 레고로_자르기(text):
    # ① 문장을 조각으로 자르고 각 조각을 번호로 바꿔요 (AI는 이 번호만 봐요)
    tokens = enc.encode(text)

    # ② 각 번호가 컴퓨터 안에서 실제로 어떤 데이터(바이트)인지 꺼내요
    raw = enc.decode_tokens_bytes(tokens)

    # ③ 잘린 조각을 하나씩 글자로 바꿔서 모아요
    pieces = []
    for b in raw:
        try:  # 조각을 글자로 바꿔 보고
            pieces.append(b.decode("utf-8"))
        except UnicodeDecodeError:  # 글자의 일부만 담긴 조각이면 <EA B0> 같은 코드로 표시해요
            pieces.append("<" + b.hex(" ").upper() + ">")

    # ④ 쪼개진 조각들을 다시 붙여서 온전한 글자로 보여줘요
    글자경계 = set()  # 글자 하나하나가 끝나는 위치를 적어둘 곳
    pos = 0
    for ch in text:
        pos += len(ch.encode("utf-8"))  # 한글은 한 글자에 3바이트, 영어는 1바이트를 차지해요
        글자경계.add(pos)

    groups = []        # 다시 붙인 글자들을 담을 곳
    묶음바이트 = b""    # 지금 붙이고 있는 조각들 (처음엔 비어 있어요)
    cnt = 0            # 지금까지 몇 조각을 붙였는지
    pos = 0            # 지금까지 읽은 위치

    for b in raw:
        묶음바이트 += b      # 조각을 하나 이어 붙이고
        cnt += 1             # 붙인 조각 수를 1 늘리고
        pos += len(b)        # 읽은 위치를 앞으로 옮겨요

        if pos in 글자경계:  # 글자가 딱 끝나는 위치에 도착하면
            s = 묶음바이트.decode("utf-8")  # 모은 조각을 글자로 바꾸고
            groups.append(f"{s}({cnt}조각)" if cnt > 1 else s)  # 여러 조각이었다면 몇 조각인지 같이 적어요
            묶음바이트, cnt = b"", 0  # 다음 글자를 위해 비워요

    # ⑤ 결과를 화면에 보여줘요
    print(f'# "{text}"  →  🧱 {len(tokens)}조각')  # 원래 문장과 총 조각 수
    print(f'#  번호: {tokens}')  # AI가 실제로 보는 번호 목록
    print(f'#  조각: {" | ".join(pieces)}')  # ③ 잘린 조각 그대로
    print(f'#  묶음: {" | ".join(groups)}')  # ④ 글자 단위로 다시 붙인 모습
    print()  # 한 줄 띄우기
```

### 게임2. 다음 단어 맞히기 (19슬라이드)
- AI 진짜 예측을 잘할까 ?
- 사람 VS AI

```
pip install openai
```

```
import math                          # 계산 도구 (확률을 %로 바꿀 때 사용)
from google.colab import userdata    # 코랩에 저장해 둔 비밀 값(API 키)을 꺼내는 도구
from openai import OpenAI            # ChatGPT를 코드에서 부를 수 있게 해주는 도구

# 1. 코랩에 저장해 둔 API 키(사용 허가증)로 ChatGPT에 연결해요
api_key = userdata.get('OPENAI_API_KEY')
client = OpenAI(api_key=api_key)


# 문장을 주면 AI가 다음에 올 단어 후보 5개를 확률과 함께 보여주고, 내 예상이 몇 등인지 알려주는 기능
def 다음단어_맞히기(prompt, 내_예상):
    # 2. ChatGPT에게 질문을 보내요
    response = client.chat.completions.create(
        model="gpt-4o-mini",  # 사용할 AI 모델
        messages=[
            # AI에게 "대화하지 말고 다음 단어 하나만 말해"라고 규칙을 알려줘요
            {"role": "system", "content": "You are a sentence completion engine. Output EXACTLY ONE word that naturally follows the user's prompt. Do NOT repeat the user's prompt, and do NOT write a full sentence."},
            {"role": "user", "content": prompt}  # 우리가 넣은 문장
        ],
        max_tokens=3,     # 딱 3조각(토큰)만 답하게 해요
        logprobs=True,    # 각 후보를 얼마나 확신하는지 점수도 같이 달라고 해요
        top_logprobs=5,   # 가장 유력한 후보 5개를 받아요
        temperature=0.5   # 답의 무작위성 (낮을수록 늘 비슷한 답)
    )

    # 3. 받은 답에서 후보 5개 목록만 꺼내요
    top_logprobs_list = response.choices[0].logprobs.content[0].top_logprobs

    print(f'\n"{prompt}" 다음에 올 단어 후보:')

    rank = -1          # 내 예상 단어의 등수 (-1은 '아직 못 찾음'이라는 뜻)
    current_rank = 1   # 지금 보고 있는 후보가 몇 등인지

    # 4. 후보를 1등부터 하나씩 보여줘요
    for logprob_obj in top_logprobs_list:
        token = logprob_obj.token       # 후보 단어
        logprob = logprob_obj.logprob   # AI가 매긴 점수 (계산용 숫자라 알아보기 어려워요)

        # 그 점수를 우리가 아는 퍼센트(%)로 바꿔요
        prob = math.exp(logprob) * 100

        # 단어 앞뒤 공백을 없애서 비교하기 쉽게 해요
        clean_word = token.strip()

        # 확률만큼 █ 막대를 그려요 (100%면 24칸)
        bar_length = int((prob / 100) * 24)
        bar = '█' * bar_length

        # 단어, 막대, 확률을 줄 맞춰 출력해요
        print(f"# {clean_word:>10}  {bar:<24} {prob:>5.1f}%")

        # 이 후보가 내 예상 단어와 같으면 등수를 기록해요 (대소문자 무시)
        if clean_word.lower() == 내_예상.lower() and rank ==-1:
            rank = current_rank

        current_rank += 1  # 다음 후보는 한 등 아래

    # 5. 내 예상 단어가 몇 등이었는지 알려줘요
    if rank != -1:  # 후보 5개 안에 있으면
        print(f"# 👉 '{내_예상}' 은(는) AI 표에서 {rank}등!")
    else:  # 후보 5개 안에 없으면
        print(f"# 👉 '{내_예상}' 은(는) 상위 5개 후보에 없습니다.")

    # AI가 실제로 쓴 전체 답변을 보여줘요
    print("AI 답변:", response.choices[0].message.content)
```

### 게임3. 누가 더 가까울까 (21슬라이드)
- 어떻게 예측이 가능할까?
- 좌표 세상 알아보기

```
import numpy as np                   # 숫자 계산 도구 (좌표끼리 더하고 빼고 비교할 때 사용)
from openai import OpenAI            # ChatGPT 회사(OpenAI)의 AI를 코드에서 부를 수 있게 해주는 도구
from google.colab import userdata    # 코랩에 저장해 둔 비밀 값(API 키)을 꺼내는 도구

# 1. 코랩에 저장해 둔 API 키(사용 허가증)로 OpenAI에 연결해요
api_key = userdata.get('OPENAI_API_KEY')
client = OpenAI(api_key=api_key)


# 단어를 넣으면 그 뜻을 담은 좌표(숫자 1536개)로 바꿔줘요 (뜻이 비슷한 단어는 좌표도 가까워요)
def 좌표(text):
    response = client.embeddings.create(
        input=text,                       # 좌표로 바꿀 단어
        model="text-embedding-3-small"    # 단어를 좌표로 바꿔주는 AI 모델
    )
    return response.data[0].embedding     # 받은 결과에서 좌표만 꺼내 돌려줘요


# 두 좌표가 얼마나 같은 방향을 가리키는지 점수로 계산해요 (1에 가까울수록 뜻이 비슷해요)
def 코사인_유사도(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))


# 결과를 █ 막대그래프로 보기 좋게 출력해요
def 출력_포맷팅(제목, 결과_리스트):
    print(f'# "{제목}" 에 가까운 순서:')
    for cand, sim in 결과_리스트:
        bar_len = max(0, int(sim * 30))   # 점수만큼 막대 길이를 정해요 (1점이면 30칸, 음수면 0칸)
        bar = "█" * bar_len               # 그 길이만큼 █를 그려요
        print(f'#    {bar:<30} {sim:6.2f}  {cand}')  # 막대, 점수, 단어를 줄 맞춰 출력해요
    print()  # 한 줄 띄우기


# 기준 단어 하나와 후보 단어들을 주면, 뜻이 가까운 순서대로 줄 세워 보여줘요
def 누가_더_가까울까(target, candidates):
    target_vec = 좌표(target)             # 기준 단어를 좌표로 바꿔요

    results = []                          # (단어, 점수)를 모아둘 곳
    for cand in candidates:
        cand_vec = 좌표(cand)                         # 후보 단어를 좌표로 바꾸고
        sim = 코사인_유사도(target_vec, cand_vec)      # 기준 단어와 얼마나 비슷한지 점수를 매겨요
        results.append((cand, sim))                   # 결과를 모아둬요

    results.sort(key=lambda x: x[1], reverse=True)    # 점수가 높은 순서대로 줄 세워요
    출력_포맷팅(target, results)                       # 막대그래프로 보여줘요


# 단어끼리 더하고 빼서(예: 왕 - 남자 + 여자) 나온 좌표가 어떤 후보 단어와 가장 가까운지 보여줘요
def 단어_산술(w1, w2, w3, candidates):
    v1, v2, v3 = np.array(좌표(w1)), np.array(좌표(w2)), np.array(좌표(w3))  # 세 단어를 좌표로 바꿔요
    target_vec = v1 - v2 + v3             # 좌표끼리 빼고 더해서 새 좌표를 만들어요

    results = []                          # (단어, 점수)를 모아둘 곳
    for cand in candidates:
        cand_vec = np.array(좌표(cand))               # 후보 단어를 좌표로 바꾸고
        sim = 코사인_유사도(target_vec, cand_vec)      # 계산해서 나온 좌표와 얼마나 비슷한지 점수를 매겨요
        results.append((cand, sim))                   # 결과를 모아둬요

    results.sort(key=lambda x: x[1], reverse=True)    # 점수가 높은 순서대로 줄 세워요
    출력_포맷팅(f"{w1} - {w2} + {w3}", results)        # 막대그래프로 보여줘요
```

### 게임4. 눈치 게임 (26슬라이드)
- 단어의 뜻이 하나로 고정되어 있을까?
- 대화 속에서 단어 이해하기

```
import math                          # 계산 도구 (확률을 %로 바꿀 때 사용)
from google.colab import userdata    # 코랩에 저장해 둔 비밀 값(API 키)을 꺼내는 도구
from openai import OpenAI            # ChatGPT를 코드에서 부를 수 있게 해주는 도구

# 1. 코랩에 저장해 둔 API 키(사용 허가증)로 ChatGPT에 연결해요
api_key = userdata.get('OPENAI_API_KEY')
client = OpenAI(api_key=api_key)


# 조각(예: bib)을 주면 AI에게 물어서 완성된 단어(예: bibimbap)로 바꿔줘요
def 단어_완성(prompt, 조각):
    response = client.chat.completions.create(
        model="gpt-4o-mini",  # 사용할 AI 모델
        messages=[
            # AI에게 "문장 다음에 올, 이 조각으로 시작하는 단어를 완성해서 그 단어만 말해"라고 알려줘요
            {"role": "system", "content": "You complete a single word. The next word after the given sentence starts with the given letters. Output ONLY that complete word. If the letters are already a complete word, output them as they are."},
            {"role": "user", "content": f"Sentence: {prompt}\nStarts with: {조각}"}  # 문장과 조각을 함께 보내요
        ],
        max_tokens=10,    # 긴 단어도 잘리지 않게 10조각까지 허용해요
        temperature=0     # 무작위성 0 (매번 똑같은 답이 나와요)
    )
    return response.choices[0].message.content.strip(" .\n")  # 앞뒤 공백·마침표를 없애고 단어만 돌려줘요


# 문장을 주면 AI가 다음에 올 단어 후보 5개를 확률과 함께 보여주는 기능
def 다음단어_후보(prompt):
    # 2. ChatGPT에게 질문을 보내요
    response = client.chat.completions.create(
        model="gpt-4o-mini",  # 사용할 AI 모델
        messages=[
            # AI에게 "대화하지 말고 다음 단어 하나만 말해"라고 규칙을 알려줘요
            {"role": "system", "content": "You are a sentence completion engine. Output EXACTLY ONE word that naturally follows the user's prompt. Do NOT repeat the prompt."},
            {"role": "user", "content": prompt}  # 우리가 넣은 문장
        ],
        max_tokens=1,       # 후보 목록은 첫 조각 기준이라 1조각이면 충분해요
        logprobs=True,      # 각 후보를 얼마나 확신하는지 점수도 같이 달라고 해요
        top_logprobs=10,    # 가장 유력한 후보 10개를 받아요 (최대 20개까지 가능)
        temperature=0       # 무작위성 0 (매번 똑같은 답이 나와요)
    )

    # 3. 받은 답에서 후보 10개 목록만 꺼내요
    top_candidates = response.choices[0].logprobs.content[0].top_logprobs

    display_count = 0  # 지금까지 화면에 보여준 후보 수

    print(f'"{prompt}" 다음 단어 후보:')

    # 4. 후보를 1등부터 하나씩 확인하며 보여줘요
    for candidate in top_candidates:
        token_str = candidate.token      # 후보 조각
        logprob = candidate.logprob      # AI가 매긴 점수 (계산용 숫자라 알아보기 어려워요)

        clean_token = token_str.strip()  # 조각 앞뒤 공백을 없애요

        # 공백이나 줄바꿈뿐인 빈 후보는 건너뛰어요
        if not clean_token:
            continue

        # 5개를 다 보여줬으면 멈춰요
        if display_count >= 5:
            break

        prob = math.exp(logprob) * 100            # 점수를 우리가 아는 퍼센트(%)로 바꿔요
        word = 단어_완성(prompt, clean_token)      # 조각을 완성된 단어로 바꿔요
        print(f"    {word:<12} {prob:5.1f}% ")  # 완성 단어, 확률, 원래 조각을 출력해요

        display_count += 1  # 보여준 후보 수를 1 늘려요
    print()  # 한 줄 띄우기
```

### 게임5. 왜 이 단어? (29슬라이드)
- 문장으로의 완성은 어떻게 되는걸까?

```
import math                          # 계산 도구 (확률을 %로 바꿀 때 사용)
from google.colab import userdata    # 코랩에 저장해 둔 비밀 값(API 키)을 꺼내는 도구
from openai import OpenAI            # ChatGPT를 코드에서 부를 수 있게 해주는 도구

# 1. 코랩에 저장해 둔 API 키(사용 허가증)로 ChatGPT에 연결해요
api_key = userdata.get('OPENAI_API_KEY')
client = OpenAI(api_key=api_key)


# AI가 글을 한 조각씩 쓸 때, 매번 어떤 후보 중에서 주사위를 굴려 골랐는지 단계별로 보여주는 기능
# 단어수 = 몇 조각까지 쓸지, 온도 = 얼마나 모험적으로 고를지 (높을수록 1등이 아닌 후보도 잘 뽑혀요)
def 왜_이_단어를_골랐나(prompt, 단어수=6, 온도=0.8):
    print(f"🎲 AI가 문장을 한 번에 {단어수}토큰만큼 생성하고, 그 과정을 분석합니다... (현재 온도: {온도})\n")

    # 2. ChatGPT에게 이야기를 이어 쓰게 해요 (딱 1번만 부르고, 쓴 과정을 전부 받아와요)
    response = client.chat.completions.create(
        model="gpt-4o-mini",  # 사용할 AI 모델
        messages=[
            {"role": "system", "content": "You are a text completion engine. Continue the user's story naturally."},  # "이야기를 자연스럽게 이어 써"라는 규칙
            {"role": "user", "content": prompt}  # 우리가 넣은 문장
        ],
        max_tokens=단어수,    # 이만큼의 조각(토큰)까지만 쓰게 해요
        temperature=온도,     # 우리가 정한 온도로 주사위를 굴리게 해요
        logprobs=True,        # 각 후보를 얼마나 확신하는지 점수도 같이 달라고 해요
        top_logprobs=4        # 매 조각마다 가장 유력한 후보 4개를 받아요
    )

    current_text = prompt  # 지금까지 완성된 문장 (처음엔 우리가 넣은 문장)
    tokens_info = response.choices[0].logprobs.content  # AI가 쓴 조각들과 그때의 후보 기록 전체

    # 3. 조각을 하나씩 다시 재생하며, 매번 어떻게 골랐는지 보여줘요
    for step, token_info in enumerate(tokens_info, start=1):
        chosen_token = token_info.token        # 이번 단계에 실제로 뽑힌 조각
        top_candidates = token_info.top_logprobs  # 이번 단계의 후보 4개

        # 첫 조각 앞에 띄어쓰기가 빠지는 경우가 많아서, 보기 좋게 공백 1칸을 넣어줘요
        if step == 1 and not current_text.endswith(" ") and chosen_token and chosen_token[0].isalpha():
            current_text += " "

        print(f"# ━━ {step}단계 ━━ 지금까지: {current_text} ▮")

        cumulative = 0.0   # 주사위 눈금을 어디까지 채웠는지 (0에서 시작해 확률만큼 늘어나요)
        chosen_rank = -1   # 뽑힌 조각이 후보 중 몇 등인지 (-1은 '후보 4개 안에 없음')
        rank = 0           # 지금 보고 있는 후보의 등수 (0이 1등)

        top1_token = top_candidates[0].token  # 1등 후보 (나중에 비교할 때 써요)

        # 4. 후보마다 확률만큼 주사위 칸을 나눠서 보여줘요
        for candidate in top_candidates:
            token_str = candidate.token      # 후보 조각
            logprob = candidate.logprob      # AI가 매긴 점수 (계산용 숫자라 알아보기 어려워요)

            prob = math.exp(logprob) * 100   # 점수를 우리가 아는 퍼센트(%)로 바꿔요
            start_range = cumulative         # 이 후보 칸이 시작하는 눈금
            end_range = cumulative + prob    # 이 후보 칸이 끝나는 눈금 (확률이 클수록 칸이 넓어요)
            cumulative = end_range           # 다음 후보는 여기서부터 칸을 채워요

            marker = " ◀ 뽑힘" if token_str == chosen_token else ""  # 실제로 뽑힌 후보면 표시를 달아요
            if token_str == chosen_token:
                chosen_rank = rank           # 뽑힌 후보의 등수를 기록해요

            print(f"#     {token_str!r:>8} {prob:5.1f}%   주사위 눈금 {start_range:5.1f} ~ {end_range:5.1f}{marker}")  # 후보, 확률, 주사위 칸을 출력해요
            rank += 1  # 다음 후보는 한 등 아래

        # 뽑힌 조각이 후보 4개 안에 없었다면, 확률 낮은 후보가 뽑혔다고 알려줘요
        if chosen_rank == -1:
            print(f"#     {chosen_token!r:>8}   ... (그 외 낮은 확률의 단어가 운 좋게 뽑힘) ◀ 뽑힘")

        # 5. 왜 이 조각이 뽑혔는지 한 줄로 설명해요
        if chosen_rank == 0:  # 1등이 뽑혔으면
            print(f"#   → {chosen_token!r} 이 뽑힘. 왜? 1등이라서 — 가장 넓은 칸에 주사위가 떨어짐")
        else:  # 1등이 아닌 후보가 뽑혔으면
            print(f"#   → 🎲 {chosen_token!r} 이 뽑힘. 왜? 1등은 {top1_token!r} 이었지만 주사위가 {chosen_token!r} 칸에 떨어짐")

        print()  # 한 줄 띄우기

        current_text += chosen_token  # 뽑힌 조각을 문장 뒤에 붙이고 다음 단계로 넘어가요
```

```
# 함수 실행
왜_이_단어를_골랐나("Once upon a time, a teacher", 단어수=6, 온도=1.6)
```

### 게임6. 조종하고 숙이기 (34슬라이드)
- 우리는 어떻게 AI를 다루어야 할까 ?


```
import math                          # 계산 도구 (확률을 %로 바꿀 때 사용)
from google.colab import userdata    # 코랩에 저장해 둔 비밀 값(API 키)을 꺼내는 도구
from openai import OpenAI            # ChatGPT를 코드에서 부를 수 있게 해주는 도구

# 1. 코랩에 저장해 둔 API 키(사용 허가증)로 ChatGPT에 연결해요
api_key = userdata.get('OPENAI_API_KEY')
client = OpenAI(api_key=api_key)


# 조각(예: po)을 주면 AI에게 물어서 완성된 단어(예: poor)로 바꿔줘요
def 단어_완성(prompt, 조각):
    response = client.chat.completions.create(
        model="gpt-4o-mini",  # 사용할 AI 모델
        messages=[
            # AI에게 "문장 다음에 올, 이 조각으로 시작하는 단어를 완성해서 그 단어만 말해"라고 알려줘요
            {"role": "system", "content": "You complete a single word. The next word after the given sentence starts with the given letters. Output ONLY that complete word. If the letters are already a complete word, output them as they are."},
            {"role": "user", "content": f"Sentence: {prompt}\nStarts with: {조각}"}  # 문장과 조각을 함께 보내요
        ],
        max_tokens=10,    # 긴 단어도 잘리지 않게 10조각까지 허용해요
        temperature=0     # 무작위성 0 (매번 똑같은 답이 나와요)
    )
    return response.choices[0].message.content.strip(" .\n")  # 앞뒤 공백·마침표를 없애고 단어만 돌려줘요


# 문장을 주면 AI가 다음에 올 쉬운 단어 후보 5개를 확률과 함께 보여주는 기능 (temp = 온도)
def 다음단어_후보(prompt, temp=0.0):
    print(f'입력 문장: "{prompt}"')

    # 2. ChatGPT에게 질문을 보내요
    response = client.chat.completions.create(
        model="gpt-4o-mini",  # 사용할 AI 모델
        messages=[
            {
                "role": "system",
                "content":  "You are a sentence completion engine. Output EXACTLY ONE word that naturally follows the user's prompt. Do NOT repeat the prompt."
            },
            {"role": "user", "content": prompt}  # 우리가 넣은 문장
        ],
        max_tokens=1,       # 후보 목록은 첫 조각 기준이라 1조각이면 충분해요
        temperature=temp,   # 우리가 정한 온도로 답을 고르게 해요
        logprobs=True,      # 각 후보를 얼마나 확신하는지 점수도 같이 달라고 해요
        top_logprobs=5      # 가장 유력한 후보 5개를 받아요
    )

    # 3. 받은 답에서 후보 5개 목록만 꺼내요
    top_candidates = response.choices[0].logprobs.content[0].top_logprobs

    # 4. 후보를 1등부터 하나씩 완성된 단어로 바꿔서 보여줘요
    print("단어 후보 및 확률:")
    for candidate in top_candidates:
        token = candidate.token          # 후보 조각
        logprob = candidate.logprob      # AI가 매긴 점수 (계산용 숫자라 알아보기 어려워요)
        prob = math.exp(logprob) * 100   # 점수를 우리가 아는 퍼센트(%)로 바꿔요

        조각 = token.strip()                                  # 조각 앞뒤 공백을 없애요
        word = 단어_완성(prompt, 조각) if 조각 else token      # 조각을 완성된 단어로 바꿔요 (빈 조각은 그대로 둬요)

        # 완성된 단어, 확률, 원래 조각을 출력해요 (조각은 앞 공백까지 보이게 따옴표로 감싸요)
        print(f"  {word:>10} : {prob:>5.1f}%")

    print("-" * 50)  # 구분선 긋기
```

```
# STEER
# 엄격한 교수
다음단어_후보("You are a strict professor. Feedback: Your essay is")
```

```
# 따뜻한 튜터
다음단어_후보("You are a warm tutor. Feedback: Your essay is")
```
